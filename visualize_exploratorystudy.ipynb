```python
# POS lookup
WF_POS = [
    wave_slowUpHardDrop,
    wave_expRiseLinFall,
    wave_impulseDampedTail,
    wave_asymHalfSine,
    wave_amBurstAsym,
    wave_quadPushLinReturn
]

# NEG are polarity-reversed
WF_NEG = [[-v for v in arr] for arr in WF_POS]

# Code mapping 0..11
code_rows = []
for code in range(12):
    wf = code >> 1         # 0..5
    neg = code & 1         # 0=+, 1=-
    name = ["slowUpHardDrop","expRiseLinFall","impulseDampedTail","asymHalfSine","amBurstAsym","quadPushLinReturn"][wf]
    sign = "+" if neg == 0 else "-"
    code_rows.append({"code": code, "waveform": name, "sign": sign})

df = pd.DataFrame(code_rows)
from caas_jupyter_tools import display_dataframe_to_user
display_dataframe_to_user("Code→Waveform mapping (0..11)", df)

# Plot each code on its own chart
for code in range(12):
    wf = code >> 1
    neg = code & 1
    y = WF_NEG[wf] if neg else WF_POS[wf]
    x = list(range(len(y)))
    plt.figure(figsize=(8, 3))
    plt.plot(x, y)
    title = f"Code {code}  ·  {['slowUpHardDrop','expRiseLinFall','impulseDampedTail','asymHalfSine','amBurstAsym','quadPushLinReturn'][wf]}  ({'NEG' if neg else 'POS'})"
    plt.title(title)
    plt.xlabel("Sample index (0..255)")
    plt.ylabel("Amplitude (int16)")
    plt.tight_layout()
    plt.show()
```
